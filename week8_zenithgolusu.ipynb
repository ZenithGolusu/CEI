# 🧠 Week 8 - Agentic AI Assignment
## Single Agent Pipeline Project

This notebook implements a simple Agentic AI pipeline with:
- Calculator Tool
- Keyword Extraction Tool
- Conditional Routing
- Structured JSON Output
- Interactive Testing


```python
# TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
```

```python
# TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words=text.split()
        keywords=list(set([w.lower().strip(".,!?") for w in words if len(w)>4]))
        return keywords[:5]
    except Exception:
        return []
```

```python
# AGENT

def agent(query:str):
    query_lower=query.lower()

    if "calculate" in query_lower:
        expr=query_lower.replace("calculate","").strip()
        return {
            "type":"calculation",
            "result":calculator(expr)
        }

    elif "keyword" in query_lower or "keywords" in query_lower:
        text=query
        text=text.replace("Extract keywords from","")
        text=text.replace("extract keywords from","")
        return {
            "type":"keywords",
            "result":extract_keywords(text)
        }

    else:
        return {
            "type":"general",
            "result":"Machine learning is a branch of Artificial Intelligence that enables computers to learn from data without being explicitly programmed."
        }
```

```python
# TEST CASES

queries=[
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:",q)
    print("Response:",agent(q))
    print("-"*60)
```

```python
# INTERACTIVE MODE

while True:
    user=input("Enter query (type 'exit' to stop): ")

    if user.lower()=="exit":
        print("Goodbye!")
        break

    print(agent(user))
```
